## Dataset Splitting with `train_test_split`

### Why do we split our dataset into Training and Testing sets?**
**Answer:**
If we train a model on our entire dataset, we cannot verify how it performs on new, unseen data. It might just memorize the training dataset (a problem known as **Overfitting**).


### How does `train_test_split` work? Show the code for splitting `iphone.csv` with a test size of 0.33.**
We separate our features ($X$, inputs) and targets ($y$, output/label) first, then split them using a test size of 0.33.

```python
x_iphone=df_iphone[['number']]
y_iphone=df_iphone['Price']
```

```python
x_iphone
```

```python
y_iphone
```

### Show exactly what `X_train`, `X_test`, `y_train`, and `y_test` look like. Explain why specific rows ended up in which set.**
Run the following cell to inspect the resulting training and testing divisions.

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test = train_test_split(x_iphone,y_iphone,test_size=0.33,random_state=42)
#the use of random_state
# sequence of the x_train,...... should be same
```

```python
print("--- X_train (Features for Training, 4 rows) ---")
print(x_train)
print("\n--- y_train (Labels for Training, 4 rows) ---")
print(y_train)

print("\n--- X_test (Features for Training, 2 rows) ---")
print(x_test)
print("\n--- y_test (Labels for Training, 2 rows) ---")
print(y_test)
```

## Simple Linear Regression Part -2 

## Step 1. Import the necessary libraries

```python
import numpy as np
import pandas as pd
```

## Step 2. Import the dataset

```python
df=pd.read_csv("50_Startups.csv",sep=',')
```

```python
df.head()
```

## Step 3 . Check the State Column

```python
df['State'].value_counts()
```

## Step 4 . Splitting dataset in to input and output

```python
x=df[['R&D Spend','Administration','Marketing Spend','State']]
x
```

## Step 5 . Convert state Column into Numeric Column

## Step 5.1 .  Perform Transformation

```python
x1=pd.get_dummies(x,columns=['State'],drop_first=True)
```

```python
x1
```

## Step 6 . Dummy variable trap

```python
# Already Performed using | drop_first =True
x1=x1.astype(int)
x1
```

## Step 7 Splitting dataset in to Train and Test

```python
y=df['Profit']
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x1,y,train_size=0.8,random_state=42)
```

## Step 8  Import LinearRegression model from linear_model family 

```python
from sklearn.linear_model import LinearRegression
model=LinearRegression()
```

## Step 9  Fit the data

```python
model.fit(x_train,y_train)
```

## Step 10  Predict the data

```python
y_predict=model.predict(x_test)
```
