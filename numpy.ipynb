```python
import time
import random
import numpy as np


N = 300

list_A = [[random.random() for _ in range(N)] for _ in range(N)]
list_B = [[random.random() for _ in range(N)] for _ in range(N)]

array_A = np.array(list_A)
array_B = np.array(list_B)


print(f"Benchmarking {N}x{N} Matrix Multiplication...\n")

start_time = time.time()
list_result = [[0.0 for _ in range(N)] for _ in range(N)]
for i in range(N):
    for j in range(N):
        for k in range(N):
            list_result[i][j] += list_A[i][k] * list_B[k][j]
python_time = time.time() - start_time
print(f"Pure Python List Time: {python_time:.4f} seconds")



start_time = time.time()
array_result = array_A @ array_B  
numpy_time = time.time() - start_time
print(f"NumPy Array Time: {numpy_time:.4f} seconds")


speedup = python_time / numpy_time
print(f"\n NumPy is {speedup:.1f}x FASTER than standard Python lists!")
```

```python
import requests
import bs4
```

```python
req = requests.get("https://darshan.ac.in/faculty-list/btech-computer")
```

```python
soup = bs4.BeautifulSoup(req.text,"lxml")
```

```python
data = soup.select("body > main > div:nth-child(5) > div > div > div > div > div")

for f in data:
    name = f.select("h2")
    e = f.select("span")
    print(name[0].text.strip() + " -->" +e[3].text.strip())
```
